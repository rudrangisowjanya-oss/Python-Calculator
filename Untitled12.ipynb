```python
# ============================================
# AI & ML Basics using NumPy, Pandas & Matplotlib
# Google Colab Ready Program
# ============================================

# Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================
# Create Sample Dataset
# ============================================

student_data = {
    "Name": ["Rahul", "Priya", "Anil", "Sneha", "Kiran"],
    "Age": [20, 21, 19, 20, 22],
    "Marks": [85, 90, 78, 88, 95]
}

data = pd.DataFrame(student_data)

print("===== STUDENT DATASET =====")
print(data)

# ============================================
# NumPy
# ============================================

print("\n===== NUMPY =====")

marks_array = np.array(data["Marks"])

print("Marks Array:", marks_array)
print("Sum:", np.sum(marks_array))
print("Average:", np.mean(marks_array))
print("Maximum:", np.max(marks_array))
print("Minimum:", np.min(marks_array))

# ============================================
# Pandas - Explore Dataset
# ============================================

print("\n===== FIRST 5 ROWS =====")
print(data.head())

print("\n===== LAST 5 ROWS =====")
print(data.tail())

print("\n===== DATASET INFO =====")
print(data.info())

print("\n===== DATASET DESCRIPTION =====")
print(data.describe())

# ============================================
# Data Cleaning
# ============================================

print("\n===== MISSING VALUES =====")
print(data.isnull().sum())

data = data.drop_duplicates()
data = data.dropna()

print("\nData Cleaned Successfully!")

# ============================================
# Simple Data Analysis
# ============================================

print("\n===== DATA ANALYSIS =====")

print("Average Marks :", data["Marks"].mean())
print("Highest Marks :", data["Marks"].max())
print("Lowest Marks  :", data["Marks"].min())

# ============================================
# Bar Chart
# ============================================

plt.figure(figsize=(6,4))
plt.bar(data["Name"], data["Marks"])
plt.title("Student Marks - Bar Chart")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.show()

# ============================================
# Line Chart
# ============================================

plt.figure(figsize=(6,4))
plt.plot(data["Name"], data["Marks"], marker='o')
plt.title("Student Marks - Line Chart")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.show()

# ============================================
# Pie Chart
# ============================================

plt.figure(figsize=(6,6))
plt.pie(
    data["Marks"],
    labels=data["Name"],
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Student Marks Distribution")
plt.show()

print("\n===== PROGRAM COMPLETED SUCCESSFULLY =====")
```
